### Instructions for getting dataset from Kaggle directly to Datahub

Go to https://www.kaggle.com/settings

Under **API** `Create Token`

Upload the `kaggle.json` file to datahub, move it to `~/.kaggle`

You can run the following command in the terminal for this step.

`mkdir -p ~/.kaggle`

`mv /path/to/kaggle.json ~/.kaggle/` The path is where you uploaded the json.

Run this command in the terminal:

`chmod 600 ~/.kaggle/kaggle.json` It gives only you read/write permissions

You can then run the following notebook.

```python
#DOCUMENTATION
#https://github.com/Kaggle/kaggle-api/blob/main/kaggle/api/kaggle_api_extended.py
from kaggle.api.kaggle_api_extended import KaggleApi

api = KaggleApi()
api.authenticate()
```

```python
kaggle_url = 'vamlau/smaller-dataset'
destination_path = './Dataset'

files = api.dataset_list_files(kaggle_url).files
file_names = [f.name for f in files]

print(file_names)
```

```python
###Not recommended to upload all at once, I ran into a disk quota exceeded error
# api.dataset_download_files(url, path=destination_path, unzip=True)

#18 total files
for i, f in enumerate(file_names):
    api.dataset_download_file(kaggle_url, file_name=f, path=destination_path)
    print(f'{i + 1}. Collected file: {f}')
```
